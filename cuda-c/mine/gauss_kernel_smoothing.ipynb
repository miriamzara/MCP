```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
N = 100
x = np.linspace(0, np.pi, N)
y = np.sin(x) + np.random.random(N)*0.1

plt.scatter(x, y)
plt.plot(x, y)

radius = 3
kernel = np.array([0.004, 0.054, 0.242, 0.399, 0.242, 0.054, 0.004])
y_augmented = np.concatenate((np.zeros(radius), y, np.zeros(radius)))
y_smooth = np.zeros(N)
for i in range(N):
    i_augmented = i + radius
    y_slice = y_augmented[(i_augmented-radius):(i_augmented+radius+1)]
    y_smooth[i] = np.dot(y_slice, kernel)

plt.scatter(x, y_smooth)
plt.plot(x, y_smooth)
```
